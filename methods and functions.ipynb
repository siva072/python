```python
#2.this is multiplicating numbers


def multiplication(*number):
    total = 1
    for num in number:
        total = total*num
    return total
```

```python
multiplication(1,2,-4,4)
```

```python
#3.this is cheking numbers in range


def ran_check(num,low,high):
    if num in range(low,high):
        print(f'{num} is range between {low} and {high}')
    else:
        print('not in range')
    
```

```python
ran_check(10,1,10)
```

```python
#4.this is counting lowercase and upper case characters


def count_low_high(s):
    lowercase=0
    uppercase=0
    for i in s:
        if i.isupper():
            uppercase +=1
        elif i.islower():
            lowercase +=1
        else:
            pass
    print(s)
    print(lowercase)
    print(uppercase)
```

```python
u = "Hi my Name is Siva Krishna Please!"
```

```python
count_low_high(u)
```

```python
#5. this is checking a word is a palindrome

def palindrome(s):
    s = s.replace(' ','')
    return s == s[::-1]
    
```

```python
palindrome('hi   h')
```
