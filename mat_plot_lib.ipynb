```python
from matplotlib import pyplot as plt
a = ['siva','tulasi']
b= [100,160]
```

```python
plt.plot(a,b)
plt.title('info')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
```

```python
plt.bar(a,b, color = 'r')
plt.title('info')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
```

```python
plt.barh(a,b)
plt.title('info')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
```
