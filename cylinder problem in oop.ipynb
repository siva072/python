```python
class Line:

    def __init__(self,co_or1,co_or2):
        self.co_or1= co_or1
        self.co_or2= co_or2

    def distance(self):
        x1,y1 = self.co_or1
        x2,y2 = self.co_or2
        return ((x2-x1)**2 + (y2-y1)**2)**0.5

    def slope(self):
        x1,y1 =self.co_or1
        x2,y2 =self.co_or2
        return ((y2-y1)/(x2-x1))
```

```python
class Cylinder:

     pi = 3.14
    
     def __init__(self,radius,height):
        self.radius = radius
        self.height = height

     def volume(self):
        return Cylinder.pi * self.radius **2 * self.height

     def surface_area(self):
        return 2*Cylinder.pi*self.radius*self.height
```

```python
c1=Cylinder(5,10)
```

```python
c1.volume()
```
