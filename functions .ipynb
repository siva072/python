```python
def myfunc(a,b):
    if a%2==0 and b%2==0:
        return min(a,b)
    elif a%2==1 or b%2==1:
        return max(a,b)
```

```python
myfunc(10,20)
```

```python
myfunc(12,69)
```

```python
def animal(text):
    for i in text.split():
        if i[0]==i[0][1]:
            return True
        else:
            return False
    return False        
```

```python
animal('lion ing nbhv')
```

```python
animal('kangaroo aus')
```

```python
def old(name):
    text = name[:3]
    test = name[3:]
    return text.capitalize() + test.capitalize()
```

```python
old('sivakrishnas')
```

```python
def sum(a,b):
    if a+b==20:
        return True
    elif a==20 or b==20:
        return True
    else:
        return False
```

```python
def reverse(text):
    te = text.split()
    reverse_list=te[::-1] 
    return ''.join(reverse_list)
```

```python
reverse('ting tonvhjb')
```

```python
def checkimg(*nums):
    for i in range(0,len(nums)-1):
      if nums[i]== 3 and nums[i+1] ==3:
         return True
    return False      
```

```python
checkimg(2,43,4,2,3,3,4)
```

```python
def blackjack(a,b,c):
    if a+b+c <= 21:
        return a+b+c
    elif a+b+c >21 and a==11 or b==11 or c==11:
        return (a+b+c)-10
    else:
        return 'BUST'
    
```

```python
blackjack(12,10,11)
```
